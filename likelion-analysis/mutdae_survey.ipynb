```python
import pandas as pd

# 파일 경로 및 시트명
file_path = r"C:\Users\manid\Desktop\data_study\mutdae_survey_2504.xlsx"
sheet_name = "bBkxg1cd"

# 엑셀 데이터 불러오기
df = pd.read_excel(file_path, sheet_name=sheet_name)

# 데이터프레임 기본 정보 확인
print("행 수:", df.shape[0])
print("열 수:", df.shape[1])
df.head(3)  # 상위 3행 미리보기
```

```python
#'*전공이 IT 개발 관련 전공인가요?*' 란 질문의 응답자별로  '어떤 학습을 진행하고 있는지 알려주세요.' 응답 통계를 산출해.
import pandas as pd

# [2] melt 대상 컬럼 그룹 지정
learning_cols = [
    "그룹 스터디", "학교 프로그램", "유료 인터넷 강의", "무료 인터넷 강의", "오프라인 학원", "무료 온라인 부트캠프", "Other"
]

# [3] melt 수행: 어떤 학습을 진행하고 있나요?
melted = df[learning_cols + ['ID']].melt(
    id_vars='ID',
    value_name='응답여부',
    var_name='응답항목'
).dropna(subset=['응답여부'])

melted['질문'] = "어떤 학습을 진행하고 있나요?"
melted.rename(columns={"ID": "응답자ID"}, inplace=True)

# [4] 전공 여부 연결
major_map = df[['ID', '*전공이 IT 개발 관련 전공인가요?*']].rename(
    columns={'ID': '응답자ID', '*전공이 IT 개발 관련 전공인가요?*': 'IT전공여부'}
)
merged = pd.merge(melted, major_map, on='응답자ID', how='left')

# [5] pivot 수행: 응답항목 기준, 전공여부별 집계
pivot = pd.pivot_table(
    merged,
    index='응답항목',
    columns='IT전공여부',
    aggfunc='size',
    fill_value=0
).reset_index()

# ✅ [6] 행별 합계 컬럼 추가
pivot["합계"] = pivot.iloc[:, 1:].sum(axis=1)

# 결과 출력
pivot
```
