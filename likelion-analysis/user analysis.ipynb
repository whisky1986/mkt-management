```python
import pandas as pd

# 엑셀 파일 경로
file_path = r"C:\Users\manid\Desktop\데이터분석\cleaned_1Q2024_marketing_data.xlsx"

# 엑셀 파일 읽기
data = pd.read_excel(file_path, engine='openpyxl')

# 공란을 결측값으로 처리
data.replace("", pd.NA, inplace=True)

# 'createdAt1' 컬럼을 datetime 형식으로 변환
data['createdAt1'] = pd.to_datetime(data['createdAt1'], errors='coerce')

# 2024.01.01 이후 데이터 필터링
filtered_data = data[data['createdAt1'] >= '2024-01-01']

# profileUtmSource 컬럼의 고유값 확인
unique_utm_sources = filtered_data['profileUtmSource'].unique()

# 고유값 출력
print(unique_utm_sources)
```

```python
# 'createdAt1' 컬럼을 datetime 형식으로 변환
data['createdAt1'] = pd.to_datetime(data['createdAt1'], errors='coerce')

# 2024.01.01 이후 데이터 필터링
filtered_data = data[data['createdAt1'] >= '2024-01-01']

# profileUtmSource 컬럼의 고유값 확인
unique_utm_sources = filtered_data['profileUtmSource'].unique()
print("Unique profileUtmSource values:", unique_utm_sources)

# 각 스쿨별로 어떤 채널을 통해 진입했는지 분석
school_channel_analysis = filtered_data.groupby(['schoolName', 'profileUtmSource']).size().reset_index(name='count')

# 결과를 피벗 테이블 형식으로 변환하여 보기 좋게 출력
pivot_table = school_channel_analysis.pivot(index='schoolName', columns='profileUtmSource', values='count').fillna(0)

# 결과 확인
print(pivot_table)
```

```python
import pandas as pd

# 엑셀 파일 경로 및 파일명
file_path = r"C:\Users\manid\Desktop\데이터분석\cleaned_1Q2024_marketing_data - 직접분석.xlsx"

# 엑셀 파일의 시트 이름 확인
excel_file = pd.ExcelFile(file_path)
print("Available sheet names:", excel_file.sheet_names)

# 'raw' 시트가 있는지 확인하고 없으면 첫 번째 시트로 변경
sheet_name = 'raw'
if sheet_name not in excel_file.sheet_names:
    sheet_name = excel_file.sheet_names[0]
    print(f"'raw' 시트가 없어서 '{sheet_name}' 시트를 사용합니다.")

# 엑셀 파일 읽기
df = pd.read_excel(file_path, sheet_name=sheet_name, engine='openpyxl')

# birthDate 컬럼을 datetime 형식으로 변환
df['birthDate'] = pd.to_datetime(df['birthDate'], errors='coerce')

# 변환 결과 확인
print(df['birthDate'].head())

# 엑셀 파일 저장
with pd.ExcelWriter(file_path, engine='openpyxl', mode='a', if_sheet_exists='replace') as writer:
    df.to_excel(writer, sheet_name=sheet_name, index=False)

print(f"변경된 파일이 저장되었습니다: {file_path}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 데이터 준비
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'],
    'Male': [897, 906, 881, 813, 792, 710, 732, 739, 769, 708, 722, 732],
    'Female': [832, 756, 773, 780, 740, 687, 706, 751, 733, 630, 642, 649]
}

df = pd.DataFrame(data)

# 막대 그래프 그리기
plt.figure(figsize=(12, 8))
bar_width = 0.35
index = range(len(df['Month']))

bar1 = plt.bar(index, df['Male'], bar_width, label='Male', color='skyblue')
bar2 = plt.bar([i + bar_width for i in index], df['Female'], bar_width, label='Female', color='lightcoral')

plt.xlabel('Month')
plt.ylabel('Number of Birthdays')
plt.title('Birthdays by Month and Gender')
plt.xticks([i + bar_width / 2 for i in index], df['Month'])
plt.legend()

# 데이터 레이블 추가
for bar in bar1:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width() / 2 - 0.1, yval + 50, int(yval), va='center')

for bar in bar2:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width() / 2 - 0.1, yval + 50, int(yval), va='center')

plt.tight_layout()

# 그래프 저장
plt.savefig('/mnt/data/birthdays_by_month_and_gender.png')
plt.show()
```
