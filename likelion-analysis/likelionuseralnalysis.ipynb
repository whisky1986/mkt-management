```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. '가입일시'를 날짜 형식으로 변환
df_user["가입일시"] = pd.to_datetime(df_user["가입일시"], errors="coerce")

# 2. '가입연월' 컬럼 추가 (연-월 단위)
df_user["가입연월"] = df_user["가입일시"].dt.to_period("M")

# 3. 나이 및 성별 데이터 필터링 (95세 이상 & 결측치 제거)
df_filtered = df_user[
    (df_user["나이"].notna()) & 
    (df_user["나이"] < 95) & 
    (df_user["성별"].notna())
]

# 4. 가입연월을 문자열로 변환 (YYYY-MM 형식) 후 오름차순 정렬
df_filtered["가입연월"] = df_filtered["가입연월"].astype(str)
df_filtered = df_filtered.sort_values(by="가입연월")

# 5. 월별 나이 통계 (성별 포함) 계산
monthly_gender_stats = df_filtered.groupby(["가입연월", "성별"])["나이"].describe()[["25%", "50%", "75%"]]
monthly_gender_stats = monthly_gender_stats.reset_index()

# 6. 박스 플롯 시각화 (성별 구분)
plt.figure(figsize=(14, 6))
ax = sns.boxplot(data=df_filtered, x="가입연월", y="나이", hue="성별")

# 7. 레이블 추가 (Q1, Median, Q3)
for i, row in monthly_gender_stats.iterrows():
    xpos = list(df_filtered["가입연월"].unique()).index(row["가입연월"])  # X축 위치
    ypos_q1 = row["25%"]  # 1사분위수 (Q1)
    ypos_median = row["50%"]  # 중위수 (Median)
    ypos_q3 = row["75%"]  # 3사분위수 (Q3)
    
    # 레이블 추가
    plt.text(xpos, ypos_q1, f'{ypos_q1:.1f}', ha='center', fontsize=9, color='blue')
    plt.text(xpos, ypos_q3, f'{ypos_q3:.1f}', ha='center', fontsize=9, color='red')

# 8. X축을 YYYY-MM 형식으로 설정
plt.xticks(rotation=45, ha="right")
plt.xlabel("Registration Month")
plt.ylabel("Age")
plt.title("Monthly Age Distribution by Gender (Box Plot)")
plt.legend(title="Gender")

# 9. 격자 추가
plt.grid(axis="y", linestyle="--", alpha=0.7)

# 10. 그래프 출력
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import ace_tools as tools  # Jupyter에서 테이블을 보기 위한 도구

# 1. '가입일시'를 날짜 형식으로 변환
df_user["가입일시"] = pd.to_datetime(df_user["가입일시"], errors="coerce")

# 2. '가입연월' 컬럼 추가 (연-월 단위)
df_user["가입연월"] = df_user["가입일시"].dt.to_period("M")

# 3. 광고성 정보 수신 동의 데이터 필터링 (결측치 제거)
df_filtered = df_user[df_user["광고성 정보 수신 동의"].notna()]

# 4. 월별 '동의' 및 '비동의' 개수 계산
monthly_consent = df_filtered.groupby(["가입연월", "광고성 정보 수신 동의"]).size().unstack(fill_value=0)

# 5. '동의' 비율 계산 (전체 대비 %)
monthly_consent["총가입자"] = monthly_consent["동의"] + monthly_consent["비동의"]
monthly_consent["동의율(%)"] = (monthly_consent["동의"] / monthly_consent["총가입자"]) * 100

# 6. 데이터 정렬 (연월 오름차순)
monthly_consent = monthly_consent.sort_index()

# 7. 테이블 출력
tools.display_dataframe_to_user(name="월별 광고성 정보 수신 동의율", dataframe=monthly_consent)

# 8. 시각화: 동의율 변화 추이 그래프 (레이블 포함)
plt.figure(figsize=(12, 6))
plt.plot(monthly_consent.index.astype(str), monthly_consent["동의율(%)"], marker='o', linestyle='-', color='blue', label="Consent Rate (%)")

# 레이블 추가 (데이터 포인트 위에 동의율 표시)
for i, (month, rate) in enumerate(zip(monthly_consent.index.astype(str), monthly_consent["동의율(%)"])):
    plt.text(i, rate, f'{rate:.1f}%', ha='center', va='bottom', fontsize=10, color='black')

# 그래프 설정
plt.xlabel("Registration Month")
plt.ylabel("Consent Rate (%)")
plt.title("Monthly Advertising Consent Rate")
plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.legend()

# 9. 그래프 출력
plt.show()
```
