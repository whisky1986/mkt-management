```python
import pandas as pd

# 파일 경로 (본인의 경로에 맞게 수정해야 함)
file_path = r"C:\Users\manid\Downloads\applyuser_sorting.xlsx"

# 엑셀 파일 불러오기
df = pd.read_excel(file_path)

# '전공' 컬럼을 문자열로 변환하여 안전하게 처리
df["전공"] = df["전공"].astype(str)

# IT 개발자로 진로 선택이 가능한 전공(전공생)과 비전공생을 구분할 기준 설정
it_majors = [
    "컴퓨터", "컴퓨터공학", "컴퓨터과학", "소프트웨어공학", "IT공학", "소프트웨어학부",
    "정보통신공학", "정보통신공학과", "정보통계학", "데이터사이언스", "AI", "응용통계학", "통계학",
    "산업공학", "전기공학", "전자공학", "기계공학", "소프트웨어융합학과", "응용 컴퓨터공학과", "산업시스템공학과"
]

non_it_majors = [
    "경영학", "경제학", "국제학", "관광경영", "신문방송학", "디자인", "법학", "의류학", "청각학",
    "생명공학", "생명과학", "간호", "식품영양학", "의생명공학", "요리", "건축", "자동차", "광고PR브랜딩", "공간디자인학"
]

# '전공' 컬럼을 기준으로 전공생/비전공생을 구분하는 새로운 컬럼 생성
def categorize_major(major):
    if major in ["nan", "None", "없음"]:  # 빈 값 처리
        return "미입력"
    for it_major in it_majors:
        if it_major in major:
            return "전공생"
    for non_it_major in non_it_majors:
        if non_it_major in major:
            return "비전공생"
    return "기타"

df["전공여부"] = df["전공"].apply(categorize_major)

# 조건 1: [약관 동의] 값이 '1'
condition_terms_agreed = df["약관 동의"] == 1

# 조건 2: [지원상태] 값이 '지원중'
condition_application_status = df["지원상태"] == "지원중"

# 조건 3: 전공생 그룹
condition_major = df["전공여부"] == "전공생"

# 조건 4: 비전공자 중 '경영' 키워드 포함
condition_non_major_business = (df["전공여부"] == "비전공생") & (df["전공"].str.contains("경영", na=False))

# 조건 5: 기타 그룹 중 '경영' 키워드 포함
condition_other_business = (df["전공여부"] == "기타") & (df["전공"].str.contains("경영", na=False))

# 최종 조건 적용 (AND 관계)
filtered_df = df[
    condition_terms_agreed & 
    condition_application_status & 
    (condition_major | condition_non_major_business | condition_other_business)
]

# 결과 출력
print("=== 필터링된 타겟 세그먼트 데이터 ===")
print(filtered_df.head())

# 엑셀 파일로 저장
output_path = r"C:\Users\manid\Downloads\filtered_target_segment.xlsx"
filtered_df.to_excel(output_path, index=False)

print(f"\n📥 필터링된 데이터가 엑셀 파일로 저장되었습니다: {output_path}")
```

```python
import pandas as pd

# 기존 필터링된 데이터 파일 (중복 제거된 버전)
filtered_file_path = r"C:\Users\manid\Downloads\filtered_target_segment_deduplicated.xlsx"

# 프론트엔드 및 백엔드 지원자 명단 파일 경로
frontend_applicants_path = r"C:\Users\manid\Downloads\feb apply_list_2025-03-12.csv"
backend_applicants_path = r"C:\Users\manid\Downloads\beb apply_list_2025-03-12.csv"

# 데이터 불러오기
filtered_df = pd.read_excel(filtered_file_path)
frontend_df = pd.read_csv(frontend_applicants_path)
backend_df = pd.read_csv(backend_applicants_path)

# '연락처' 컬럼을 문자열로 변환 (데이터 비교를 위해)
filtered_df["연락처"] = filtered_df["연락처"].astype(str)
frontend_df["연락처"] = frontend_df["연락처"].astype(str)
backend_df["연락처"] = backend_df["연락처"].astype(str)

# 현재 지원 중인 지원자 목록 생성
current_applicants = pd.concat([frontend_df["연락처"], backend_df["연락처"]]).drop_duplicates()

# 현재 지원 중인 지원자를 제외한 데이터 필터링
final_filtered_df = filtered_df[~filtered_df["연락처"].isin(current_applicants)]

# 결과 출력
print("=== 현재 지원자를 제외한 최종 필터링된 타겟 세그먼트 데이터 ===")
print(final_filtered_df.head())

# 최종 엑셀 파일 저장
final_output_path = r"C:\Users\manid\Downloads\final_filtered_target_segment.xlsx"
final_filtered_df.to_excel(final_output_path, index=False)

print(f"\n📥 현재 지원자를 제외한 최종 데이터가 엑셀 파일로 저장되었습니다: {final_output_path}")
```
